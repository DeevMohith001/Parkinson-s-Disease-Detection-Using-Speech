Read all Excel sheets


```python

import pandas as pd

file_path = "Demographics_age_sex.xlsx"

# Load every sheet from the original Excel workbook
sheets = pd.read_excel(file_path, sheet_name=None)

print("Number of sheets:", len(sheets))
print("\nSheet names:")
for name, df in sheets.items():
    print(f"{name}: {df.shape[0]} rows × {df.shape[1]} columns")

# Inspect the main demographics/diagnosis sheet
display(sheets["Parselmouth"].head())
```

Merge feature sheets

```python

import pandas as pd

demo = sheets["Parselmouth"].copy()
demo.columns = demo.columns.astype(str).str.strip()

# Standardize the participant ID in the main sheet
demo = demo.rename(columns={"Sample": "Sample ID"})
df = demo.copy()

feature_sheets = [
    "LPC_means", "LAR_means", "Cep_means", "MFCC_means",
    "LPC_vars", "LAR_vars", "Cep_vars", "MFCC_vars"
]

for name in feature_sheets:
    feature_df = sheets[name].copy()
    feature_df.columns = feature_df.columns.astype(str).str.strip()

    # Feature sheets use 'Sample' as the participant ID
    if "Sample" in feature_df.columns:
        feature_df = feature_df.rename(columns={"Sample": "Sample ID"})
    elif "Sample ID" not in feature_df.columns:
        raise ValueError(f"ID column not found in {name}: {feature_df.columns.tolist()}")

    # Remove repeated label; the main sheet is the source of truth
    if "Label" in feature_df.columns:
        feature_df = feature_df.drop(columns=["Label"])

    # Clean IDs before merging
    feature_df["Sample ID"] = feature_df["Sample ID"].astype(str).str.strip()
    df["Sample ID"] = df["Sample ID"].astype(str).str.strip()

    df = df.merge(
        feature_df,
        on="Sample ID",
        how="inner",
        validate="one_to_one"
    )

# Validate the merged dataset
print("Merged dataset shape:", df.shape)
print("Unique participants:", df["Sample ID"].nunique())
print("Duplicate IDs:", df["Sample ID"].duplicated().sum())
print("Missing values:", df.isnull().sum().sum())

print("\nDiagnosis distribution:")
print(df["Label"].value_counts())

display(df.head())
```

Build clean dataset and define X, y

```python

# Remove duplicate suffixes from feature names
# Keep the first occurrence if the same feature name appears more than once.
# We will verify duplicates before dropping anything.

print("Duplicate column names:", df.columns[df.columns.duplicated()].tolist())

# The expected 80 speech features should have unique names.
# Rebuild clean feature columns directly from the original sheets.
clean_df = sheets["Parselmouth"].copy()
clean_df.columns = clean_df.columns.astype(str).str.strip()

if "Sample" in clean_df.columns:
    clean_df = clean_df.rename(columns={"Sample": "Sample ID"})

for name in feature_sheets:
    feature_df = sheets[name].copy()
    feature_df.columns = feature_df.columns.astype(str).str.strip()

    if "Sample" in feature_df.columns:
        feature_df = feature_df.rename(columns={"Sample": "Sample ID"})

    # Keep participant ID and actual numeric feature columns only
    feature_cols = [
        col for col in feature_df.columns
        if col not in ["Sample ID", "Label"]
    ]
    feature_df = feature_df[["Sample ID"] + feature_cols]

    clean_df = clean_df.merge(
        feature_df,
        on="Sample ID",
        how="inner",
        validate="one_to_one"
    )

df = clean_df

# Encode labels: Parkinson's = 1, Healthy = 0
df["Target"] = df["Label"].map({"PwPD": 1, "HC": 0})

# Separate speech features from demographic/identifier columns
exclude_cols = ["Sample ID", "Label", "Age", "Sex", "Target"]
feature_cols = [c for c in df.columns if c not in exclude_cols]

X = df[feature_cols].apply(pd.to_numeric, errors="raise")
y = df["Target"]

print("Clean dataset shape:", df.shape)
print("Number of speech features:", len(feature_cols))
print("Missing values in X:", X.isnull().sum().sum())
print("\nTarget distribution:")
print(y.value_counts().rename(index={0: "Healthy", 1: "Parkinson's"}))

display(X.head())
```

Compare baseline models

```python

import numpy as np
import pandas as pd

from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import SelectKBest, f_classif

from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis

# Reproducibility
SEED = 42

# Confirm dataset is ready
assert X.shape == (81, 80), f"Unexpected X shape: {X.shape}"
assert y.nunique() == 2, "Expected two diagnosis classes."
assert not X.isnull().any().any(), "X contains missing values."

# 5-fold stratified CV: preserves class proportions in each fold
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)

# Feature selection happens inside each training fold to reduce leakage
k_features = 30

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=5000, class_weight="balanced", random_state=SEED
    ),
    "SVM (RBF)": SVC(
        kernel="rbf", C=1.0, class_weight="balanced",
        probability=True, random_state=SEED
    ),
    "SVM (Linear)": SVC(
        kernel="linear", C=1.0, class_weight="balanced",
        probability=True, random_state=SEED
    ),
    "KNN": KNeighborsClassifier(n_neighbors=5),
    "Random Forest": RandomForestClassifier(
        n_estimators=300, class_weight="balanced",
        random_state=SEED
    ),
    "Extra Trees": ExtraTreesClassifier(
        n_estimators=300, class_weight="balanced",
        random_state=SEED
    ),
    "LDA": LinearDiscriminantAnalysis(solver="lsqr", shrinkage="auto")
}

scoring = {
    "accuracy": "accuracy",
    "balanced_accuracy": "balanced_accuracy",
    "precision": "precision",
    "recall": "recall",
    "f1": "f1",
    "roc_auc": "roc_auc"
}

results = []
fitted_pipelines = {}

for name, model in models.items():
    pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("selector", SelectKBest(score_func=f_classif, k=k_features)),
        ("scaler", StandardScaler()),
        ("model", model)
    ])

    scores = cross_validate(
        pipeline,
        X,
        y,
        cv=cv,
        scoring=scoring,
        return_train_score=False,
        n_jobs=-1,
        error_score="raise"
    )

    results.append({
        "Model": name,
        "Accuracy_mean": scores["test_accuracy"].mean(),
        "Accuracy_std": scores["test_accuracy"].std(),
        "Balanced_Accuracy": scores["test_balanced_accuracy"].mean(),
        "Precision": scores["test_precision"].mean(),
        "Recall": scores["test_recall"].mean(),
        "F1": scores["test_f1"].mean(),
        "ROC_AUC": scores["test_roc_auc"].mean()
    })

    fitted_pipelines[name] = pipeline

results_df = pd.DataFrame(results).sort_values(
    by="Balanced_Accuracy", ascending=False
).reset_index(drop=True)

display(results_df.round(3))
```

Tune KNN and SVM, including Yeo–Johnson

```python
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, PowerTransformer
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC

cv_tune = StratifiedKFold(
    n_splits=5, shuffle=True, random_state=42
)

pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("yeo", "passthrough"),
    ("selector", SelectKBest(score_func=f_classif)),
    ("scaler", StandardScaler()),
    ("model", KNeighborsClassifier())
])

yeo_options = [
    "passthrough",
    PowerTransformer(method="yeo-johnson", standardize=False)
]

param_grid = [
    {
        "yeo": yeo_options,
        "selector__k": [10, 20, 30, 40, 60, "all"],
        "model": [KNeighborsClassifier()],
        "model__n_neighbors": [3, 5, 7, 9, 11],
        "model__weights": ["uniform", "distance"],
        "model__p": [1, 2]
    },
    {
        "yeo": yeo_options,
        "selector__k": [10, 20, 30, 40, 60, "all"],
        "model": [SVC()],
        "model__kernel": ["rbf"],
        "model__C": [0.1, 1, 10, 100],
        "model__gamma": ["scale", 0.01, 0.1],
        "model__class_weight": [None, "balanced"]
    }
]

search = GridSearchCV(
    estimator=pipeline,
    param_grid=param_grid,
    scoring="balanced_accuracy",
    cv=cv_tune,
    n_jobs=-1,
    refit=True,
    return_train_score=False
)

search.fit(X, y)

print("Best CV balanced accuracy:", round(search.best_score_, 3))
print("\nBest parameters:")
print(search.best_params_)

best_model = search.best_estimator_
```

Nested Cross-Validation for Robust Model Evaluation

```python
import numpy as np
from sklearn.model_selection import (
    StratifiedKFold, GridSearchCV, cross_validate
)

outer_cv = StratifiedKFold(
    n_splits=5, shuffle=True, random_state=123
)

inner_cv = StratifiedKFold(
    n_splits=4, shuffle=True, random_state=42
)

nested_search = GridSearchCV(
    estimator=pipeline,
    param_grid=param_grid,
    scoring="balanced_accuracy",
    cv=inner_cv,
    n_jobs=-1,
    refit=True
)

nested_results = cross_validate(
    nested_search,
    X,
    y,
    cv=outer_cv,
    scoring={
        "accuracy": "accuracy",
        "balanced_accuracy": "balanced_accuracy",
        "precision": "precision",
        "recall": "recall",
        "f1": "f1",
        "roc_auc": "roc_auc"
    },
    n_jobs=1
)

for metric in nested_results:
    if metric.startswith("test_"):
        scores = nested_results[metric]
        print(
            f"{metric.replace('test_', '').upper()}: "
            f"{scores.mean():.3f} ± {scores.std():.3f}"
        )
```

```python
import pandas as pd
from sklearn.model_selection import GridSearchCV, cross_validate

scoring = {
    "accuracy": "accuracy",
    "balanced_accuracy": "balanced_accuracy",
    "precision": "precision",
    "recall": "recall",
    "f1": "f1",
    "roc_auc": "roc_auc"
}

def evaluate_model(X_data, model_name):
    search = GridSearchCV(
        estimator=pipeline,
        param_grid=param_grid,
        scoring="balanced_accuracy",
        cv=inner_cv,
        n_jobs=-1,
        refit=True
    )

    results = cross_validate(
        search,
        X_data,
        y,
        cv=outer_cv,
        scoring=scoring,
        n_jobs=1
    )

    return {
        "Model": model_name,
        **{
            metric: f"{results['test_' + metric].mean():.3f} ± "
                    f"{results['test_' + metric].std():.3f}"
            for metric in scoring
        }
    }

comparison = pd.DataFrame([
    evaluate_model(X, "Speech Only"),
    evaluate_model(X_demo, "Speech + Age + Sex")
])

display(comparison)
```

corrected comparison

```python
import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, PowerTransformer
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.model_selection import GridSearchCV, cross_validate
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC

# Acoustic feature columns
speech_cols = list(X.columns)

# Preprocessing for speech features only
speech_preprocessor = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("yeo", PowerTransformer(method="yeo-johnson", standardize=False)),
    ("selector", SelectKBest(score_func=f_classif)),
    ("scaler", StandardScaler())
])

# Demographic preprocessing
demo_preprocessor = ColumnTransformer([
    ("age", StandardScaler(), ["Age"]),
    ("sex", "passthrough", ["Sex"])
])

# Model A: Speech only
features_A = ColumnTransformer([
    ("speech", speech_preprocessor, speech_cols)
])

pipeline_A = Pipeline([
    ("features", features_A),
    ("model", SVC())
])

# Model B: Speech + Age + Sex
features_B = ColumnTransformer([
    ("speech", speech_preprocessor, speech_cols),
    ("demo", demo_preprocessor, ["Age", "Sex"])
])

pipeline_B = Pipeline([
    ("features", features_B),
    ("model", SVC())
])

# Hyperparameter grids
param_grid_A = [
    {
        "features__speech__selector__k": [10, 20, 30, 40, 60, "all"],
        "model": [KNeighborsClassifier()],
        "model__n_neighbors": [3, 5, 7, 9, 11],
        "model__weights": ["uniform", "distance"],
        "model__p": [1, 2]
    },
    {
        "features__speech__selector__k": [10, 20, 30, 40, 60, "all"],
        "model": [SVC()],
        "model__kernel": ["rbf"],
        "model__C": [0.1, 1, 10, 100],
        "model__gamma": ["scale", 0.01, 0.1],
        "model__class_weight": [None, "balanced"]
    }
]

param_grid_B = [
    {
        "features__speech__selector__k": [10, 20, 30, 40, 60, "all"],
        "model": [KNeighborsClassifier()],
        "model__n_neighbors": [3, 5, 7, 9, 11],
        "model__weights": ["uniform", "distance"],
        "model__p": [1, 2]
    },
    {
        "features__speech__selector__k": [10, 20, 30, 40, 60, "all"],
        "model": [SVC()],
        "model__kernel": ["rbf"],
        "model__C": [0.1, 1, 10, 100],
        "model__gamma": ["scale", 0.01, 0.1],
        "model__class_weight": [None, "balanced"]
    }
]

scoring = {
    "accuracy": "accuracy",
    "balanced_accuracy": "balanced_accuracy",
    "precision": "precision",
    "recall": "recall",
    "f1": "f1",
    "roc_auc": "roc_auc"
}

def nested_evaluate(estimator, param_grid, X_data):
    search = GridSearchCV(
        estimator,
        param_grid,
        scoring="balanced_accuracy",
        cv=inner_cv,
        n_jobs=-1
    )

    return cross_validate(
        search,
        X_data,
        y,
        cv=outer_cv,
        scoring=scoring,
        n_jobs=1
    )

results_A = nested_evaluate(pipeline_A, param_grid_A, X)
results_B = nested_evaluate(pipeline_B, param_grid_B, X_demo)

comparison = []

for name, results in [
    ("Speech Only", results_A),
    ("Speech + Age + Sex", results_B)
]:
    row = {"Model": name}
    for metric in scoring:
        scores = results["test_" + metric]
        row[metric] = f"{scores.mean():.3f} ± {scores.std():.3f}"
    comparison.append(row)

display(pd.DataFrame(comparison).set_index("Model"))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import ConfusionMatrixDisplay, confusion_matrix

def get_oof_predictions(estimator, param_grid, X_data):
    predictions = np.zeros(len(y))

    for train_idx, test_idx in outer_cv.split(X_data, y):
        search = GridSearchCV(
            estimator,
            param_grid,
            scoring="balanced_accuracy",
            cv=inner_cv,
            n_jobs=-1
        )

        search.fit(X_data.iloc[train_idx], y[train_idx])

        predictions[test_idx] = search.predict(
            X_data.iloc[test_idx]
        )

    return predictions

# Generate predictions
pred_A = get_oof_predictions(pipeline_A, param_grid_A, X)
pred_B = get_oof_predictions(pipeline_B, param_grid_B, X_demo)

# Plot confusion matrices
fig, axes = plt.subplots(1, 2, figsize=(11, 4))

for ax, pred, name in [
    (axes[0], pred_A, "Speech Only"),
    (axes[1], pred_B, "Speech + Age + Sex")
]:
    ConfusionMatrixDisplay(
        confusion_matrix(y, pred),
        display_labels=["Healthy", "Parkinson's"]
    ).plot(ax=ax, cmap="Blues", colorbar=False)

    ax.set_title(name)

plt.tight_layout()
plt.show()
```

Out-of-fold predictions and confusion matrices

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

from sklearn.model_selection import GridSearchCV
from sklearn.metrics import (
    confusion_matrix,
    ConfusionMatrixDisplay,
    classification_report
)

def get_oof_predictions(estimator, param_grid, X_data):
    predictions = np.zeros(len(y), dtype=int)

    for train_idx, test_idx in outer_cv.split(X_data, y):
        search = GridSearchCV(
            estimator=estimator,
            param_grid=param_grid,
            scoring="balanced_accuracy",
            cv=inner_cv,
            n_jobs=-1
        )

        search.fit(X_data.iloc[train_idx], y[train_idx])
        predictions[test_idx] = search.predict(X_data.iloc[test_idx])

    return predictions


pred_A = get_oof_predictions(pipeline_A, param_grid_A, X)
pred_B = get_oof_predictions(pipeline_B, param_grid_B, X_demo)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

for ax, pred, name in [
    (axes[0], pred_A, "Speech Only"),
    (axes[1], pred_B, "Speech + Age + Sex")
]:
    ConfusionMatrixDisplay(
        confusion_matrix(y, pred, labels=[0, 1]),
        display_labels=["Healthy", "Parkinson's"]
    ).plot(ax=ax, cmap="Blues", colorbar=False)

    ax.set_title(name)

plt.tight_layout()
plt.show()

print("Speech + Age + Sex classification report:")
print(classification_report(
    y, pred_B,
    target_names=["Healthy", "Parkinson's"],
    zero_division=0
))
```
